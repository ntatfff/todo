```python
# Parameters
batchSize = 1200
```

```python
# Root dataset path
rootPath = 'dataset/1219p/glcm/kernel5/'
outputPath = f'dataset/{batchSize}p/glcm/kernel5/'
numberOfTrain = 0.7 * batchSize
numberOfValidation = 0.1 * batchSize
numberOfTest = 0.2 * batchSize

trainLimit = numberOfTrain;
validationLimit = trainLimit + numberOfValidation;
testLimit = validationLimit + numberOfTest;
```

```python
import os
import pandas as pd
from pathlib import Path

datasetPath = rootPath + 'non_tumor'
train: list[pd.DataFrame] = []
validation: list[pd.DataFrame] = []
test: list[pd.DataFrame] = []
count: int = 0

patientFolders = os.listdir(datasetPath)
patientFolders.sort()
for patientFolder in patientFolders:
  if patientFolder.startswith('BraTS2021_'):
    if count < trainLimit:
      csvPath = f"{datasetPath}/{patientFolder}/nrrd2csv.csv"
      df = pd.read_csv(csvPath)
      df.insert(0, 'label', 0)
      train.append(df)
    if count >= trainLimit and count < validationLimit:
      csvPath = f"{datasetPath}/{patientFolder}/nrrd2csv.csv"
      df = pd.read_csv(csvPath)
      df.insert(0, 'label', 0)
      validation.append(df)
    if count >= validationLimit and count < testLimit:
      csvPath = f"{datasetPath}/{patientFolder}/nrrd2csv.csv"
      df = pd.read_csv(csvPath)
      df.insert(0, 'label', 0)
      test.append(df)
    count += 1

mergedTrain = pd.concat(train, ignore_index=True)
pathMergedTrain = outputPath + 'non_tumor/dataset.train.csv'
Path(pathMergedTrain).parent.mkdir(parents=True, exist_ok=True)
mergedTrain.to_csv(pathMergedTrain, index=False)

mergedValidation = pd.concat(validation, ignore_index=True)
pathMergedValidation = outputPath + 'non_tumor/dataset.validation.csv'
Path(pathMergedValidation).parent.mkdir(parents=True, exist_ok=True)
mergedValidation.to_csv(pathMergedValidation, index=False)

mergedTest = pd.concat(test, ignore_index=True)
pathMergedTest = outputPath + 'non_tumor/dataset.test.csv'
Path(pathMergedTest).parent.mkdir(parents=True, exist_ok=True)
mergedTest.to_csv(pathMergedTest, index=False)
```

```python
import pandas as pd
from pathlib import Path

# Merge tumor and non-tumor datasets for training
tumorDfTrain = pd.read_csv(outputPath + 'tumor/dataset.train.csv')
nonTumorDfTrain = pd.read_csv(outputPath + 'non_tumor/dataset.train.csv')

mergedTrain = pd.concat([tumorDfTrain, nonTumorDfTrain], ignore_index=True)
mergedTrain = mergedTrain.sample(frac=1)  # Shuffle the dataset
pathMergedTrain = outputPath + 'dataset.train.csv'
Path(pathMergedTrain).parent.mkdir(parents=True, exist_ok=True)
mergedTrain.to_csv(pathMergedTrain, index=False)

# Merge tumor and non-tumor datasets for validation
tumorDfValidation = pd.read_csv(outputPath + 'tumor/dataset.validation.csv')
nonTumorDfValidation = pd.read_csv(outputPath + 'non_tumor/dataset.validation.csv')

mergedValidation = pd.concat([tumorDfValidation, nonTumorDfValidation], ignore_index=True)
mergedValidation = mergedValidation.sample(frac=1)  # Shuffle the dataset
pathMergedValidation = outputPath + 'dataset.validation.csv'
Path(pathMergedValidation).parent.mkdir(parents=True, exist_ok=True)
mergedValidation.to_csv(pathMergedValidation, index=False)

# Merge tumor and non-tumor datasets for testing
tumorDfTest = pd.read_csv(outputPath + 'tumor/dataset.test.csv')
nonTumorDfTest = pd.read_csv(outputPath + 'non_tumor/dataset.test.csv')

mergedTest = pd.concat([tumorDfTest, nonTumorDfTest], ignore_index=True)
mergedTest = mergedTest.sample(frac=1)  # Shuffle the dataset
pathMergedTest = outputPath + 'dataset.test.csv'
Path(pathMergedTest).parent.mkdir(parents=True, exist_ok=True)
mergedTest.to_csv(pathMergedTest, index=False)
```
