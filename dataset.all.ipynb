```python
import os
import pandas as pd

listDir = os.listdir('dataset/firstorder/kernel3/non_tumor/')
listDir.sort()
for name in listDir:
  if name.startswith('BraTS2021_'):
    firstorderDf = pd.read_csv(f"dataset/firstorder/kernel3/non_tumor/{name}/nrrd2csv.csv")
    glcmDf = pd.read_csv(f"dataset/glcm/kernel3/non_tumor/{name}/nrrd2csv.csv")
    gldmDf = pd.read_csv(f"dataset/gldm/kernel3/non_tumor/{name}/nrrd2csv.csv")
    glrlmDf = pd.read_csv(f"dataset/glrlm/kernel3/non_tumor/{name}/nrrd2csv.csv")
    glszmDf = pd.read_csv(f"dataset/glszm/kernel3/non_tumor/{name}/nrrd2csv.csv")
    ngtdmDf = pd.read_csv(f"dataset/ngtdm/kernel3/non_tumor/{name}/nrrd2csv.csv")

    allDf = pd.concat([firstorderDf, glcmDf, gldmDf, glrlmDf, glszmDf, ngtdmDf], axis=1)
    os.makedirs(f"dataset/all/kernel3/non_tumor/{name}", exist_ok=True)
    allDf.to_csv(f"dataset/all/kernel3/non_tumor/{name}/nrrd2csv.csv", index=False)
print("Done")
```
