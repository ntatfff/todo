```python
kernel = 5
```

```python
import os
import pandas as pd

listDir = os.listdir(f'dataset/100p/firstorder/kernel{kernel}/non_tumor/')
listDir.sort()
for name in listDir:
  if name.startswith('BraTS2021_'):
    firstorderDf = pd.read_csv(f"dataset/100p/firstorder/kernel{kernel}/non_tumor/{name}/nrrd2csv.csv")
    glcmDf = pd.read_csv(f"dataset/100p/glcm/kernel{kernel}/non_tumor/{name}/nrrd2csv.csv")
    gldmDf = pd.read_csv(f"dataset/100p/gldm/kernel{kernel}/non_tumor/{name}/nrrd2csv.csv")
    glrlmDf = pd.read_csv(f"dataset/100p/glrlm/kernel{kernel}/non_tumor/{name}/nrrd2csv.csv")
    glszmDf = pd.read_csv(f"dataset/100p/glszm/kernel{kernel}/non_tumor/{name}/nrrd2csv.csv")
    ngtdmDf = pd.read_csv(f"dataset/100p/ngtdm/kernel{kernel}/non_tumor/{name}/nrrd2csv.csv")

    allDf = pd.concat([firstorderDf, glcmDf, gldmDf, glrlmDf, glszmDf, ngtdmDf], axis=1)
    os.makedirs(f"dataset/100p/all/kernel{kernel}/non_tumor/{name}", exist_ok=True)
    allDf.to_csv(f"dataset/100p/all/kernel{kernel}/non_tumor/{name}/nrrd2csv.csv", index=False)
print("Done")
```
