```python
# Root dataset path
rootPath = 'dataset/glszm/kernel5-radius5/'
```

```python
import pandas as pd
from pathlib import Path

tumorDf = pd.read_csv(rootPath + 'tumor/BraTS2021_00002/nrrd2csv.csv')
nonTumorDf = pd.read_csv(rootPath + 'non_tumor/BraTS2021_00002/nrrd2csv.csv')

merged = pd.concat([tumorDf, nonTumorDf], ignore_index=True)
merged = merged.sample(frac=1)  # Shuffle the dataset
output_merged = rootPath + 'dataset.1p.csv'
Path(output_merged).parent.mkdir(parents=True, exist_ok=True)
merged.to_csv(output_merged, index=False)
```
