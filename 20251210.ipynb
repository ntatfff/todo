```python
# Create a 20x20x20 array filled with 1s
import numpy as np
arr = np.ones((20, 20, 20), dtype=int)
arr1 = np.ones((20, 20, 20), dtype=int)
print(arr1[1])

# Transform: keep 1s only where x,y,z indices are divisible by (radius//2)+1
kernel = 2
diameter = kernel * 2 + 1
mask = np.zeros_like(arr, dtype=bool)
  
# Create index grids
x, y, z = np.indices(arr.shape)
mask = (x % diameter == kernel) & (y % diameter == kernel) & (z % diameter == kernel)
# Apply mask
arr = np.where(mask, 1, 0)

def andFunc(x: int, isMasked: bool) -> int:
    return int(x and isMasked)

andFuncVectorized = np.vectorize(andFunc)

result = andFuncVectorized(arr1, mask)
result[2]
```
