# Reproducibility audit — PyRadiomics

Đây **không phải thí nghiệm hiệu năng**. Chạy trong đúng môi trường/notebook đã dùng để trích xuất Radiomics.

Cách tốt nhất: đặt các cell audit ngay **sau khi đối tượng `extractor` đã được tạo và cấu hình**.

```python
import json
import radiomics

print("PyRadiomics version:", radiomics.__version__)
```

```python
settings = {}
settings['kernelRadius'] = 5
settings['maskedKernel'] = False
settings['voxelBatch'] = 1000
extractor = radiomics.featureextractor.RadiomicsFeatureExtractor(**settings)
```

## Effective settings

```python
audit = {
    "pyradiomics_version": radiomics.__version__,
    "settings": dict(extractor.settings),
    "enabledFeatures": dict(extractor.enabledFeatures),
    "enabledImagetypes": dict(extractor.enabledImagetypes)
}

print(json.dumps(audit, indent=2, default=str))

with open("pyradiomics_reproducibility_audit.json", "w", encoding="utf-8") as f:
    json.dump(audit, f, indent=2, default=str)

print("Saved: pyradiomics_reproducibility_audit.json")
```
