```python
# Utilities: loaders and validation
import os
import sys
import glob
import csv
from typing import List, Tuple
import numpy as np
import SimpleITK as sitk

def load_array(path: str) -> np.ndarray:
    ext = os.path.splitext(path)[1].lower()
    if ext != '.nrrd':
        raise ValueError(f"Unsupported file type for this notebook (expected .nrrd): {path}")
    image = sitk.ReadImage(path)
    arr = sitk.GetArrayFromImage(image).astype(np.float32)
    if arr.ndim != 3:
        raise ValueError(f"Expected 3D array, got shape {arr.shape} for {path}")
    return arr

def validate_same_shape(arrs: List[np.ndarray]) -> Tuple[int, int, int]:
    shapes = [a.shape for a in arrs]
    if len(set(shapes)) != 1:
        raise ValueError(f"All arrays must have the same shape, got: {shapes}")
    return arrs[0].shape
```

```python
# Combine and export to CSV (skip all-zero rows) using pandas
import pandas as pd
import os
from pathlib import Path 

def nrrd2csv(input_files: List[str], mask: np.ndarray, output_csv: str):
  arrays = [load_array(p)[mask == 1] for p in input_files]
  validate_same_shape(arrays)
  N = len(arrays)

  # Stack as (N, X, Y, Z) then reshape to (voxels, N)
  stacked = np.stack(arrays, axis=0)  # (N, D, H, W)
  vox_mat = stacked.reshape(N, -1).T       # (D*H*W, N)


  # Create DataFrame and write CSV
  # Column names derived from final token before extension in filename
  # Example: original_firstorder_10Percentile.nrrd -> 10Percentile
  base_names = [os.path.splitext(os.path.basename(p))[0] for p in input_files]
  cols = [bn.split('_')[-1] if '_' in bn else bn for bn in base_names]
  df = pd.DataFrame(vox_mat, columns=cols)

  Path(output_csv).parent.mkdir(parents=True, exist_ok=True)
  df.to_csv(output_csv, index=False)
  print(f"Saved CSV to: {output_csv} with columns: {cols}")
```

```python
datasetPath = rootPath + 'tumor'

patientFolders = os.listdir(datasetPath)
patientFolders.sort()
for patientFolder in patientFolders:
  if patientFolder.startswith('BraTS2021_'):
    patientPath = f"{datasetPath}/{patientFolder}"
    featureFiles = list(filter(lambda f: f.endswith('.nrrd'), os.listdir(patientPath)))
    featureFiles.sort()
    featureFiles = [f"{patientPath}/{featureFile}" for featureFile in featureFiles]
    outputCsvPath = f"{patientPath}/nrrd2csv.csv"
    if os.path.exists(outputCsvPath):
      print(f"CSV already exists for {patientFolder}, skipping.")
      for featureFile in featureFiles:
        os.remove(featureFile)
      continue
    nrrd2csv(featureFiles, outputCsvPath)
    for featureFile in featureFiles:
      os.remove(featureFile)
```

```python
import os
import pandas as pd
from pathlib import Path

datasetPath = rootPath + 'non_tumor'
train: list[pd.DataFrame] = []
validation: list[pd.DataFrame] = []
test: list[pd.DataFrame] = []
count: int = 0

patientFolders = os.listdir(datasetPath)
patientFolders.sort()
for patientFolder in patientFolders:
  if patientFolder.startswith('BraTS2021_'):
    if count < 70:
      csvPath = f"{datasetPath}/{patientFolder}/nrrd2csv.csv"
      df = pd.read_csv(csvPath)
      df.insert(0, 'label', 0)
      train.append(df)
    if count >= 70 and count < 80:
      csvPath = f"{datasetPath}/{patientFolder}/nrrd2csv.csv"
      df = pd.read_csv(csvPath)
      df.insert(0, 'label', 0)
      validation.append(df)
    if count >= 80 and count < 100:
      csvPath = f"{datasetPath}/{patientFolder}/nrrd2csv.csv"
      df = pd.read_csv(csvPath)
      df.insert(0, 'label', 0)
      test.append(df)
    count += 1

mergedTrain = pd.concat(train, ignore_index=True)
pathMergedTrain = rootPath + 'non_tumor/dataset.train.csv'
Path(pathMergedTrain).parent.mkdir(parents=True, exist_ok=True)
mergedTrain.to_csv(pathMergedTrain, index=False)

mergedValidation = pd.concat(validation, ignore_index=True)
pathMergedValidation = rootPath + 'non_tumor/dataset.validation.csv'
Path(pathMergedValidation).parent.mkdir(parents=True, exist_ok=True)
mergedValidation.to_csv(pathMergedValidation, index=False)

mergedTest = pd.concat(test, ignore_index=True)
pathMergedTest = rootPath + 'non_tumor/dataset.test.csv'
Path(pathMergedTest).parent.mkdir(parents=True, exist_ok=True)
mergedTest.to_csv(pathMergedTest, index=False)
```

```python
import pandas as pd
from pathlib import Path

# Merge tumor and non-tumor datasets for training
tumorDfTrain = pd.read_csv(rootPath + 'tumor/dataset.train.csv')
nonTumorDfTrain = pd.read_csv(rootPath + 'non_tumor/dataset.train.csv')

mergedTrain = pd.concat([tumorDfTrain, nonTumorDfTrain], ignore_index=True)
mergedTrain = mergedTrain.sample(frac=1)  # Shuffle the dataset
pathMergedTrain = rootPath + 'dataset.train.csv'
Path(pathMergedTrain).parent.mkdir(parents=True, exist_ok=True)
mergedTrain.to_csv(pathMergedTrain, index=False)

# Merge tumor and non-tumor datasets for validation
tumorDfValidation = pd.read_csv(rootPath + 'tumor/dataset.validation.csv')
nonTumorDfValidation = pd.read_csv(rootPath + 'non_tumor/dataset.validation.csv')

mergedValidation = pd.concat([tumorDfValidation, nonTumorDfValidation], ignore_index=True)
mergedValidation = mergedValidation.sample(frac=1)  # Shuffle the dataset
pathMergedValidation = rootPath + 'dataset.validation.csv'
Path(pathMergedValidation).parent.mkdir(parents=True, exist_ok=True)
mergedValidation.to_csv(pathMergedValidation, index=False)

# Merge tumor and non-tumor datasets for testing
tumorDfTest = pd.read_csv(rootPath + 'tumor/dataset.test.csv')
nonTumorDfTest = pd.read_csv(rootPath + 'non_tumor/dataset.test.csv')

mergedTest = pd.concat([tumorDfTest, nonTumorDfTest], ignore_index=True)
mergedTest = mergedTest.sample(frac=1)  # Shuffle the dataset
pathMergedTest = rootPath + 'dataset.test.csv'
Path(pathMergedTest).parent.mkdir(parents=True, exist_ok=True)
mergedTest.to_csv(pathMergedTest, index=False)
```
