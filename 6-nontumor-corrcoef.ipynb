```python
def calculateCorrcoef(feature1, feature2):
    basePath = "dataset/kernel5/tumor/BraTS2021_00000"
    feature1FilePath = "%s/%s" % (basePath, feature1)
    feature2FilePath = "%s/%s" % (basePath, feature2)
    data1 = np.reshape(sitk.GetArrayFromImage(sitk.ReadImage(feature1FilePath)), -1)
    data2 = np.reshape(sitk.GetArrayFromImage(sitk.ReadImage(feature2FilePath)), -1)
    return np.corrcoef(data1, data2)[0, 1]
```

```python
offset = 1
result_df = pd.DataFrame({"feature1": [], "feature2": [], "corrcoef": []})
for i in range(len(featureNames)):
    for j in range(offset, len(featureNames)):
        result_df.loc[len(result_df)] = {
            "feature1": featureNames[i],
            "feature2": featureNames[j],
            "corrcoef": calculateCorrcoef(featureNames[i], featureNames[j]),
        }
    offset += 1

result_df.to_csv("5-tumor-corrcoef.csv")
```
