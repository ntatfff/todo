```python
import numpy as np
import nibabel as nib
from radiomics import featureextractor
import SimpleITK as sitk
```

```python
# Tạo hình ảnh giả lập 6x6 với các giá trị intensity
image_array = np.array([
    [
        [10, 12, 15, 14, 10, 12],
        [9, 11, 14, 13, 9, 11],
        [8, 10, 12, 11, 8, 10],
        [7, 9, 11, 10, 7, 9],
        [6, 8, 10, 9, 6, 8],
        [5, 7, 9, 8, 5, 7]
    ],
    [
        [10, 12, 15, 14, 10, 12],
        [9, 11, 14, 13, 9, 11],
        [8, 10, 12, 11, 8, 10],
        [7, 9, 11, 10, 7, 9],
        [6, 8, 10, 9, 6, 8],
        [5, 7, 9, 8, 5, 7]
    ],
    [
        [10, 12, 15, 14, 10, 12],
        [9, 11, 14, 13, 9, 11],
        [8, 10, 12, 11, 8, 10],
        [7, 9, 11, 10, 7, 9],
        [6, 8, 10, 9, 6, 8],
        [5, 7, 9, 8, 5, 7]
    ]
], dtype=np.float32)

# Tạo mask để chỉ định 4 pixel cần tính toán
mask_array = np.zeros_like(image_array, dtype=np.uint8)
target_pixels = [(1, 1, 1), (1, 1, 4), (1, 4, 1), (1, 4, 4)]
for x, y, z in target_pixels:
    mask_array[x, y, z] = 1  # Đánh dấu pixel trong mask

# Chuyển đổi sang định dạng NIfTI
image_nifti = nib.Nifti1Image(image_array, np.eye(4))
mask_nifti = nib.Nifti1Image(mask_array, np.eye(4))

# Lưu file tạm thời (nếu cần thiết)
nib.save(image_nifti, 'image.nii')
nib.save(mask_nifti, 'mask.nii')
```

```python
# Cấu hình Radiomics để sử dụng kernel
params = {
    'setting': {
        'kernelRadius': 1,   # Bán kính kernel
    }
}

# Tạo feature extractor
extractor = featureextractor.RadiomicsFeatureExtractor(**params)
extractor.disableAllFeatures()
extractor.enableFeatureClassByName('firstorder')
```

```python
features = extractor.execute('image.nii', 'mask.nii', voxelBased=True)

sitk.GetArrayFromImage(features['original_firstorder_Energy'])
```
