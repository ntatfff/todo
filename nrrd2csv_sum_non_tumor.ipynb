```python
import os
import pandas as pd
from pathlib import Path

datasetPath = 'dataset/firstorder/kernel5-radius5/non_tumor'
batch: list[pd.DataFrame] = []
merged_count = 0

patientFolders = os.listdir(datasetPath)
patientFolders.sort()
for patientFolder in patientFolders:
  if patientFolder.startswith('BraTS2021_'):
    csvPath = f"{datasetPath}/{patientFolder}/nrrd2csv.csv"
    df = pd.read_csv(csvPath)
    df.insert(0, 'label', 0)
    batch.append(df)
    merged_count += 1

    if merged_count % 100 == 0:
      merged = pd.concat(batch, ignore_index=True)
      output_merged = f'{datasetPath}/{merged_count}.dataset.csv'
      Path(output_merged).parent.mkdir(parents=True, exist_ok=True)
      merged.to_csv(output_merged, index=False)

if batch:
  merged = pd.concat(batch, ignore_index=True)
  output_merged = f'{datasetPath}/{merged_count}.dataset.csv'
  Path(output_merged).parent.mkdir(parents=True, exist_ok=True)
  merged.to_csv(output_merged, index=False)
```
