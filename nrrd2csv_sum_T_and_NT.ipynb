```python
import pandas as pd
from pathlib import Path

tumorDf = pd.read_csv('dataset/kernel5-radius5/tumor/dataset.csv')
nonTumorDf = pd.read_csv('dataset/kernel5-radius5/non_tumor/dataset.csv')

merged = pd.concat([tumorDf, nonTumorDf], ignore_index=True)
merged = merged.sample(frac=1)  # Shuffle the dataset
output_merged = 'dataset/kernel5-radius5/dataset.csv'
Path(output_merged).parent.mkdir(parents=True, exist_ok=True)
merged.to_csv(output_merged, index=False)
```
